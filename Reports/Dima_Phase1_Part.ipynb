# **Project Goal**

The goal of this project is to analyze student performance and study habits using data mining techniques.

For classification, the project aims to classify students into different performance categories based on their academic characteristics and study habits.

For clustering , the project aims to group students with similar academic characteristics and study habits to discover meaningful patterns in the dataset.

## Dataset Source
